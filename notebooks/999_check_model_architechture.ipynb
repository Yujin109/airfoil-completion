```python
import torch
from torchsummary import summary
```

```python
model = ConditionalResidualUNet_PosEnc()

# モデルをGPUに移す場合は以下を有効化
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model = model.to(device)

# ダミー入力データを作成
B = 4
L = 248  # 入力の長さ
x = torch.randn(B, 2, L).to(device)  # 入力波形
c = torch.randn(B, 1).to(device)  # 条件ラベル (Clなど)
t = torch.randint(0, 1000, (B,)).to(device)  # timestep（int型でもfloat型でも可）

# モデル summary の表示
summary(
    model,
    input_data=(x, c, t),
    col_names=["input_size", "output_size", "num_params", "kernel_size", "mult_adds"],
    depth=4,
)
```
